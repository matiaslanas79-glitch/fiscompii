```python
import numpy as np
rng = np.random.default_rng(seed = 42356)
#Acá se define el numero y su semilla
```

```python
L = [1.0, 2.5, 0.8, 5.0, 10.0] #Luminosidad de la estrella en unidades solares
d = [10, 25, 15, 40, 60] #Distancia de la estrella en parsecs
```

```python
def Flujo_observado(L, d):
    '''
    Esta función calcula el flujo observado de las estrellas a partir de su luminosidad y distancia.
    -----------------
    Parámetros:
    L: lista de luminosidades de las estrellas en unidades solares
    d: lista de distancias de las estrellas en parsecs
    -----------------

    Luminosidad: Arreglo de numpy para las luminosidades de las estrellas en unidades solares.
    Distancia: Arreglo de numpy para las distancias de las estrellas en parsecs.
    -----------------

    Retorna:
    Una lista con los flujos observados de las estrellas en unidades de flujo solar por parsec cuadrado.
    -----------------
    '''
    Luminosidad = np.array(L)
    Distancia = np.array(d)

    #Pasar de parsecs a metros
    d_metros = Distancia * 3.086e16 #1 parsec equivale a 3.086e16 metros

    #Calcular el flujo observado
    Flujo_observado = [Luminosidad / (4 * np.pi * d_metros**2)] #Formula dada en la tarea
    return Flujo_observado
    
```

```python
Listas_de_Flujos = Flujo_observado(L, d) 
print('Los flujos observados de las estrellas son: ')
print(*Listas_de_Flujos)
#Se calculan los flujos de las listas dadas al inicio del código.
```

```python
Maximo = np.max(Flujo_observado(L, d))
Minimo = np.min(Flujo_observado(L, d))
Promedio = np.mean(Flujo_observado(L, d))
Mediana = np.median(Flujo_observado(L, d))
#Calcula el flujo máximo, mínimo, promedio y mediano de la lista de flujos observados.

print(f'El flujo máximo observado es: {Maximo}')
print(f'El flujo mínimo observado es: {Minimo}')
print(f'El flujo promedio observado es: {Promedio}')
print(f'La mediana observada es: {Mediana}')
#Los imprime con print por temas estéticos y de claridad.
```

```python
Mascara_de_deteccion = np.array(Listas_de_Flujos) <= 10e-11
print(f'La mascara de detección es: {Mascara_de_deteccion}') #Se imprime la mascara para mostrar valores Booleanos
```

```python
indices_detectados = np.where(Mascara_de_deteccion)[0] #Se obtienen los indices de los valores que cumplen con la condición de detección.
print('Los indices de los valores que cumplen con la condición de detección son: ')
print(indices_detectados) #Se imprimen los indices de los valores que cumplen con la condición de detección.
```

```python
Lista_de_Flujo_ordenada = np.sort(Listas_de_Flujos, axis=0)[::-1] #Se ordena la lista de flujos de mayor a menor
print('La lista ordenada de mayor a menor es: ')
print(*Lista_de_Flujo_ordenada) #Se imprime la lista ordenada de mayor a menor
```

```python
import numpy as np
```

```python
tiempo = np.linspace(0, 10, 500)
#Definimos 500 puntos equidistantes entre t= 0 y t=10 años
flujo_base = np.ones(500)
#Flujo base de 500 unos
ruido = np.random.normal(loc=0, scale=0.002, size=500)
#Definimos el ruido gaussiano con media 0 y desviación estándar de 0.002
flujo = flujo_base + ruido
#Se suma el flujo base con el ruido gaussiano para obtener el flujo final

mascara_intervalo = (tiempo >= 4.5) & (tiempo <= 5.0)
#Definimos una mascara booleana para el intervalo de tiempo entre 4.5 y 5.0 años
flujo[mascara_intervalo] -= 0.02
#Le restamos 0.02 al flujo que cumple con la condición de la mascara booleana, simulando un tránsito planetario
```

```python
flujo_promedio = np.mean(flujo) #Calculamos el flujo promedio de la lista de flujos observados.
print(f'El flujo promedio es: {flujo_promedio}')
```

```python
desviacion_estandar = np.std(flujo) #Calculamos la desviación estándar de la lista de flujos observados.
print(f'La desviación estándar del flujo es: {desviacion_estandar}')
```

```python
flujo_minimo = np.min(flujo) #Se calcula el flujo mínimo de la lista de flujos observados.
index_minimo = np.argmin(flujo) #Se obtiene el índice del flujo mínimo de la lista de flujos observados.
print(f'El flujo mínimo es: {flujo_minimo} y ocurre en el índice: {index_minimo}, correspondiente al tiempo: {tiempo[index_minimo]} años')
#tiempo[index_minimo] nos da el tiempo en el que ocurre el flujo mínimo, que es el tiempo en el que ocurre el tránsito planetario.
```

```python
condicion_transito = flujo < (flujo_promedio - 3 * desviacion_estandar) #Se define la condicion de transito
flujos_detectados = flujo[condicion_transito] #Se obtienen los flujos que cumplen con la condición de tránsito

num_transitos = np.sum(condicion_transito) #Se cuenta el número de flujos que cumplen con la condición de tránsito
print(f'El número de tránsitos detectados es: {num_transitos}')
```

```python
flujo_transito = np.mean(flujos_detectados) #Se calcula el flujo promedio de los flujos que cumplen con la condición de tránsito
Delta_F = 1 - flujo_transito #Se calcula con la fórmula dada en la tarea, el valor de Delta_F
print(f'La profundidad del tránsito (Delta_F) es: {Delta_F}')
```

```python
ratio_radios = np.sqrt(Delta_F) #Se calcula la razón de radios a partir de Delta_F
print(f'La razón de radios (R_planeta/R_estrella) es: {ratio_radios}')
```

```python
tiempos_detectados = tiempo[condicion_transito] #Se obtienen los tiempos que cumplen con la condición de tránsito
tiempo_de_inicio = np.min(tiempos_detectados) #Se obtiene el tiempo de inicio del tránsito
tiempo_de_fin = np.max(tiempos_detectados) #Se obtiene el tiempo de fin del tránsito

print(f'El intervalo detectado es [{tiempo_de_inicio:.2f}, {tiempo_de_fin:.2f}] años')
print('Muy cercano al intervalo real de [4.5, 5.0] años, lo que indica que el método de detección es efectivo.')
```
